```python
from collections import Counter

customer_orders = Counter()

while True:
    product = input("Enter an ordered product (or type 'done' to finish): ").strip().lower()

    if product == "done":
        break

    if product in products:
        customer_orders[product] += 1
        print(f"{product} added to the order.")
    else:
        print("That product is not available.")
```

```python
for product, quantity in customer_orders.items():
    inventory[product] -= quantity
```

```python
def show_inventory(inventory):
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")
```

```python
customer_orders = set()

while True:
    product = input("Enter a product: ").strip().lower()

    if product in products:
        customer_orders.add(product)
        print(f"{product} added to the order.")
    else:
        print("That product is not available.")

    # Validar la respuesta yes/no
    while True:
        answer = input(
            "Do you want to add another product? (yes/no): "
        ).strip().lower()

        if answer in ("yes", "no"):
            break

        print("Please enter only 'yes' or 'no'.")

    if answer == "no":
        break
```

```python
print(customer_orders)
```

```python
total_products_ordered = len(customer_orders)
print(f"Total products ordered: {total_products_ordered}")
```

```python
percentage_ordered = (
    total_products_ordered / len(products) * 100
)

print(f"Percentage of products ordered: {percentage_ordered:.1f}%")
```
