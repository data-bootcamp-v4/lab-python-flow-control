```python
products =[ "t-shirt", "mug", "hat", "book", "keychain"]
```

```python
inventory={}
```

```python

for product in products:
    quantity = int(input("Enter the quantity for {}:".format(product)))
    inventory[product] = quantity
print(inventory)
```

```python
customer_orders=set()
x="yes"
while x== "yes":
    product = input('please enter product from these values "t-shirt", "mug", "hat", "book" or "keychain"')
    if product in products:
        customer_orders.add(product)
        x=input("please enter yes if you want to add another product" )
        inventory[product]=inventory[product]-1
    else :
         print("error")
        
       
print("done") 
print(customer_orders)
print(inventory)
```
