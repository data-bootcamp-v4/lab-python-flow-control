```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}
```

```python
for i in products:
    inventory[i] = int(input(f"How many {i} do we have?"))

inventory
```

```python
customer_orders = set()
while True:
    order = input("What would you like to order?")
    if order not in products:
         print("We don't have this product here")
    else:
        customer_orders.add(order)
        answer = input("Would you like something else? Yes or No?").lower()
        if answer == "no":
            break

customer_orders
```

```python
order_status = (len(customer_orders),round(len(customer_orders)/sum(inventory.values())*100,2))
order_status
```

```python
def order_statistics():
    print("Order Statistics:")
    print(f"Total Products Ordered: {order_status[0]}")
    print(f"Percentage of Products Ordered: {order_status[1]}%")

order_statistics()
```

```python
for product_inventory, quantity_inventory in inventory.items():
    if product_inventory in customer_orders:
        inventory[product_inventory] = quantity_inventory -1
inventory
```
