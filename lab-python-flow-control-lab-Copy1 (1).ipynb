```python
# 1. Create the list of products
products = ["t-shirt", "mug", "hat", "book", "key"]

# 2. Create an empty inventory dictionary
inventory = {}

# 3. Ask the user for the quantity of each product
for product in products:
    quantity = int(input(f"How many {product}s are available? "))
    inventory[product] = quantity
    print(inventory)

# 4. Create an empty set for customer orders
customer_orders = set()

# 5. User to enter the name of a products that a customer wants to order
while True:
    product = input("which product do you wish to order")
    customer_orders.add(product)
    x = input("anything else:")
    if x != "y":
        break

# 6. Update inventory with customer_orders
for product in customer_orders:
    if  inventory[product] > 0:
        inventory[product] -=1
    else:
        print(product,"out of stock")


print("final inventory", inventory)



















































































































        


        
        


























```
