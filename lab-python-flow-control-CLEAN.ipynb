```python
# Step 1: Start with the same products list and create inventory
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}

# Ask the user how many of each product are available
for product in products:
    quantity = int(input(f"How many '{product}' are available in inventory? "))
    inventory[product] = quantity
```

```python
# Step 2: Create a set for customer orders
customer_orders = set()

# Use a loop to keep asking for products until the user says "no"
while True:
    product = input("Enter the name of a product the customer wants to order: ").lower()

    if product in products:
        customer_orders.add(product)
        print(f"{product} added to the order.")
    else:
        print("This product is not available.")

    another = input("Do you want to add another product? (yes/no): ").lower()
    if another != "yes":
        break
```

```python
print("Customer ordered the following products:")
print(customer_orders)
```

```python
# Update inventory only for the products that were ordered
for product in customer_orders:
    if inventory[product] > 0:
        inventory[product] -= 1
    else:
        print(f"{product} is out of stock!")

# Print updated inventory
print("Updated inventory:")
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")
```
