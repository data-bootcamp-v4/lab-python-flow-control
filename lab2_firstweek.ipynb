```python
# lab2 inventory
inventory = {
    "t-shirt": 12,
    "mug": 5,
    "hat": 25,
    "book": 8
}

# Set to store customer orders
customer_orders = set()

# Loop to collect orders
while True:
    product = input("Enter the name of a product the customer wants to order: ").lower()
    customer_orders.add(product)

    another = input("Do you want to add another product? (yes/no): ").lower()
    if another != "yes":
        break

# Update inventory only for ordered products
for product in customer_orders:
    if product in inventory:
        inventory[product] -= 1

# Print results
print("\nCustomer Orders:", customer_orders)
print("Updated Inventory:", inventory)
```
