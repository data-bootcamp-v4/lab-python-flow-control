## Exercise: Managing Customer Orders Optimized

In the last lab, you were starting an online store that sells various products. To ensure smooth operations, you developed a program that manages customer orders and inventory.

You did so without using flow control. Let's go a step further and improve this code.

Follow the steps below to complete the exercise:

1. Look at your code from the lab data structures, and improve repeated code with loops.

2. Instead of asking the user to input the name of three products that a customer wants to order, do the following:
    
    a. Prompt the user to enter the name of a product that a customer wants to order.
    
    b. Add the product name to the "customer_orders" set.
    
    c. Ask the user if they want to add another product (yes/no).
    
    d. Continue the loop until the user does not want to add another product.

3. Instead of updating the inventory by subtracting 1 from the quantity of each product, only do it for the products that were ordered (those in "customer_orders").

## Solution

Here's the improved code using flow control:

```python
# Initial inventory of products
products = {
    "t-shirt": 4,
    "mug": 6,
    "hat": 9,
    "book": 3,
    "keychain": 2
}

# Initialize customer orders set
customer_orders = set()

# Step 2: Use a while loop to get customer orders
# This replaces the hardcoded approach of asking 3 times
add_more = True
while add_more:
    # Prompt user to enter a product
    product = input("Enter the name of a product that a customer wants to order: ").lower()
    
    # Add the product to the customer_orders set
    customer_orders.add(product)
    print(f"'{product}' added to the order.")
    
    # Ask if user wants to add another product
    another = input("Do you want to add another product? (yes/no): ").lower()
    
    # Continue loop based on user input
    if another != "yes":
        add_more = False

# Display customer orders
print(f"\n=== Customer Orders ===")
print(f"Products ordered: {customer_orders}")
print(f"Total unique products ordered: {len(customer_orders)}")
```

### Explanation of Step 2:

- **While Loop**: We use a `while` loop with a boolean flag `add_more` to control when to stop asking for products.
- **Dynamic Input**: Instead of asking for 3 specific products, we ask one at a time and let the user decide when to stop.
- **Set Usage**: We add each product to the `customer_orders` set. Sets automatically prevent duplicates, so if a user tries to order the same product twice, it's only stored once.
- **String Normalization**: We use `.lower()` to convert all inputs to lowercase for consistency.

```python
# Step 3: Update inventory only for the ordered products
# This uses a for loop with the customer_orders set
print("\n=== Updating Inventory ===")

for product in customer_orders:
    if product in products:
        products[product] -= 1
        print(f"Inventory updated for '{product}': {products[product]} units remaining.")
    else:
        print(f"Warning: '{product}' is not in the inventory.")

# Display updated inventory
print(f"\n=== Updated Inventory ===")
for product, quantity in products.items():
    print(f"{product.capitalize()}: {quantity} units")
```

### Explanation of Step 3:

- **For Loop Over Set**: We iterate over the `customer_orders` set using a `for` loop.
- **Conditional Check**: We use an `if` statement to verify that each ordered product exists in the inventory before updating.
- **Targeted Updates**: Unlike a previous approach that might subtract from all products, we only update the products that were actually ordered.
- **Error Handling**: If a product isn't in the inventory, we print a warning instead of causing an error.

## Key Improvements

1. **Flexible Input**: The while loop allows the user to enter any number of products, not just 3.
2. **Better Control**: We combined `while` and `for` loops with conditional statements to create a more robust program.
3. **Data Integrity**: We only update the inventory for products that were actually ordered.
4. **Error Prevention**: We validate that ordered products exist in the inventory.
5. **User-Friendly**: The program provides feedback at each step, making it clear what's happening.
