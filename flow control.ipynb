```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = {}
for item in products:
    inventory[item] =0

print("pls input available qty for each product in inventory")

for key in inventory:
    inventory[key] = int(input(f"pls input available qty for '{key}':"))

customer_orders =set()

while True:
    user_input = input(f"for the product you want to order,pls only input one product name from:'{','.join(products)}':")
    if user_input in products:
        customer_orders.add(user_input)
    else:
        user_input = input(f"product no found, pls re-input product name from:'{','.join(products)}':")
    
    if_order_more = input(f"do you want to order another product? pls reply yes or no")

    while if_order_more.lower() =='yes':
        extra_user_input = input(f"pls input extra product name you want to order from:'{','.join(products)}':")
        if extra_user_input in products:
             customer_orders.add(extra_user_input)
        else:
            user_input = input(f"product no found, pls re-input product name from:'{','.join(products)}':")
            
        if_order_more = input(f"do you want to order another product? pls reply yes or no")

    if if_order_more.lower() =='no':
        break

print("you had input all product name which you want to order, next pls input each product order qty")

customer_orders = {item:None for item in customer_orders}
for key in customer_orders:
    customer_orders[key] = int(input(f"pls input order qty for '{key}':"))

#Total_Products_Ordered = sum(customer_orders.values())
#print(f"Total Products Ordered: {Total_Products_Ordered}")

#order_product_total_available_qty = sum(inventory[key] for key in customer_orders)

#Percentage_of_Products_Ordered = (float(Total_Products_Ordered)/float(order_product_total_available_qty))*100 
#print(f"Percentage of Products Ordered: {Percentage_of_Products_Ordered:.2f}%")

#order_status = (Total_Products_Ordered,Percentage_of_Products_Ordered)

for key in customer_orders:
    inventory[key]=inventory[key]-customer_orders[key]

print("Below show updated inventory of each product after this time purchasing:")
for key,value in inventory.items():
    print(f"{key}:{value}")
```
