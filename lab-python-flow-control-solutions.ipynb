```python
products= ["t-shirt","mug","hat","book","keychain"]
inventory= {}
t_shirt=int(input("Enter the quantity of t-shirt:"))
mug=int(input("Enter the quantity of mug:"))
hat= int(input("Number of hats?"))
book= int(input("Number of books?"))
keychain= int(input("Number of keychains?"))
inventory['t-shirt']=t_shirt
inventory['mug']=mug
inventory['hat']=hat
inventory['book']=book
inventory['keychain']=keychain
```

```python
print(inventory)
customer_orders= set()
my_list=input("Which item do you want? t-shirt, mug, hat, book or keychain?").lower()
customer_orders.add(my_list)
print(customer_orders)
add_order=input("Do you want to add another product? (yes/no)").strip().lower()
add_order=(add_order=="yes")
while True:
    add_item=input("Which item do you want to add?").lower()
    customer_orders.add(add_item)
    print(f"You just added a {add_item}")
    last_add=input ("Do you want to add something more?(yes/no)").strip().lower()
    is_true=(last_add=="yes")
    #last_add se convierte en bool al hacerlo =="yes"
    if  not is_true:
        print("No more items added")
        break
        
print(customer_orders)
total_products_ordered=len(customer_orders)
print(total_products_ordered)
percentage= len(customer_orders)/len(products)*100
print( f"The percentage is {percentage}%")
order_status=(percentage,)
print(order_status)
print("Order Statistics:")
print(f"Total Products Ordered:{total_products_ordered}")
print(f"Percentage of Products Ordered:{percentage}%")

for item in customer_orders:
    if item  in inventory:
       inventory[item]-=1

for key, value in inventory.items():
    print(key,value)
```
