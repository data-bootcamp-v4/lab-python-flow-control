```python
# Exercise: Managing Customer Orders Optimized

# 1. Product list
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
# 2. Build inventory with a loop
products = ["t-shirt", "mug", "hat", "book", "keychain"]

inventory = {}
quantities = [3, 5, 2, 7, 4]

for i, product in enumerate(products):
    inventory[product] = quantities[i]

print(inventory)
```

```python
# 3. Create an empty set for customer orders
customer_orders = set()
```

```python
# 4. Ask the user to add products until they say no
while True:
    order = input("Enter the name of a product a customer wants to order: ").lower().strip()

    if order in products:
        customer_orders.add(order)
    else:
        print("This product is not available in the product list.")

    another_product = input("Do you want to add another product? (yes/no): ").lower().strip()

    if another_product == "no":
        break

```

```python
# 6. Calculate order statistics

products = ["t-shirt", "mug", "hat", "book", "keychain"]
customer_orders = {"mug", "hat", "book"}

total_products_ordered = len(customer_orders)
percentage_ordered = (total_products_ordered / len(products)) * 100
order_status = (total_products_ordered, percentage_ordered)

print("Order Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")
```

```python
# 8. Update inventory only for ordered products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

inventory = {
    "t-shirt": 10,
    "mug": 5,
    "hat": 8,
    "book": 3,
    "keychain": 12
}

customer_orders = {"mug", "hat", "book"}

for order in customer_orders:
    inventory[order] -= 5

print(inventory)
```
