```python
# defining List
products = [ "t-shirt","mug","hat","book","keychain"]
# Creating empty Dictionary
inventory = {}

# input for each product quantity
for product in products:
    quantity = int(input(f"Enter the quantity of {product}: "))
    inventory[product] = quantity
    print(f"{product} quantity is {quantity}")
print(inventory)    
```

```python
# creating empty set
customer_orders = set()
# input from the customer 
print("t-shirt","mug","hat","book","keychain")
while True:
    order = input("Enter the product you want to order: ").strip().lower()
    customer_orders.add(order)
    choice = input("Do you want to order another product? (yes/no): ").strip().lower()
    if choice != "yes":
        print("Thank you for your order!")
        break
    
print(customer_orders)
```

```python
# updaing inventory
for ele in customer_orders:
    if ele in inventory:
        inventory[ele] -= 1
print("Updated inventory:") 
print(inventory)     
```
