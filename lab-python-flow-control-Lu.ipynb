```python
# step 1: Creating empty list called products

products = ["t-shirt", "mug", "hat", "book", "keychain"] 
products
```

```python
# step 2: creating empty dictionary called inventory
inventory = {}
```

```python
# step 3a :Ask the user to input the quantity of each product available in the inventory. 
# Step 3b:Use the product names from the products list as keys in the inventory dictionary and assign the respective quantities as values. 
for product in products:
    quantity = int(input(f"how many {product} do you have? ")) # 在这个loop里先是用input function assign qty
    inventory[product] = quantity #然后这一步就是在给inventory的key赋值并不断通过循环增加键值对。赋值语句：字典名[键]=值
```

```python
print(inventory)
```

```python
#2. Instead of asking the user to input the name of three products that a customer wants to order, do the following:
#2a: Prompt the user to enter the name of a product that a customer wants to order.
#2b: Add the product name to the "customer_orders" set.
#2c: Ask the user if they want to add another product (yes/no).
#2d: Continue the loop until the user does not want to add another product.

customer_order = set() #creat an empty set called customer_order as a place holder. NOTICE: Using {} will create an empty dictionary, not a set.
customer_order.add(str(input("what product do you want to buy?")))

while True:
    answer = input("Do you want to add another prouct? Yes or No?").strip().lower()
    if answer == "no":
        print("ok thank you!")
        break
       
    
    elif answer == "yes":
        while True:
            answer = str(input("What else do you want to add?")).strip().lower()
            if answer in products:# if the answer is not in the product list
                print("ok got it!")
                customer_order.add(answer)

            elif answer == "that's it".strip().lower():
                print("All done!")
                break
            
            else:
                print("Sorry I don't have that.")
                break
               
    else:
        print("Invalid answer, please answer yes or no")

print(customer_order)
```

```python
# Step 3: Instead of updating the inventory by subtracting 1 from the quantity of each product, 
# only do it for the products that were ordered (those in "customer_orders").

for product in customer_order:
    if product in inventory:
        inventory[product] - 1 # accesing values using dictionary keys
print(inventory)
```
